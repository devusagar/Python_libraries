**Matplotlib Library**

```python
import pandas as pd
import numpy as np
import random
import matplotlib.pyplot as plt                            #Importing Matplotlib Library
```

```python
pos=[1,2,3,4,5,6,7,8,9]
run=[123,34,56,76,43,21,4,97,99]
```

```python
#Line Plot
plt.plot(pos,run)                                          #Ploting Method
plt.show()
```

```python
from matplotlib.lines import lineStyles
#Title to the graph
plt.plot(pos,run)
plt.plot(pos,run,color="red",marker='o',linestyle='--')                                       #Ploting Method
plt.title('Position Wise Scores')                          #Title for graph
```

```python
#Axis Setting
plt.plot(pos,run)                                          #Ploting Method
plt.axis([0,11,1,130])                                     #Axis Setting
```

```python
#labeling
plt.plot(pos,run)                                          #Ploting Method
plt.xlabel('Position')                                     #X_label
plt.ylabel('Runs')                                         #Y_label
```

```python
#Advance Line Plot
plt.plot(pos,run)
plt.axis([0,11,0,130])
plt.xlabel('Position',fontsize=16)
plt.ylabel('Runs',fontsize=16)
plt.title('Position Wise Scores',fontsize=20)
plt.show()
```

**Bar Plot**

```python
fields=['AI','DS','DA','ML','SE']                                      #Creating field list
Div_a=[12,13,2,4,5]
Div_b=[51,27,37,23,43]
Div_c=[14,10,4,3,2]

index=np.arange(len(fields))                                           #Creating Index for each field
```

```python
plt.bar(index,Div_a)
plt.bar(index,Div_a,color='c',width=0.2,edgecolor='r',linewidth=2)                        #Creating 1st BAR graph with styling patterns
plt.bar(index + 0.2,Div_b,color='k',width=0.2,edgecolor='r',linewidth=2)
plt.bar(index + 0.4,Div_c,color='w',width=0.2,edgecolor='r',linewidth=2)

plt.xticks(index + 0.2,fields)
plt.title('Field Wise Intrested Students')
plt.xlabel('Fields')
plt.ylabel('Students')
plt.show()

```

**Histogram Chart**

```python
RCB=np.random.randint(80,85,(100))
MI=np.random.randint(60,65,(100))
bins=[15,20,25,30,35,40,45]
```

```python
plt.hist([RCB,MI],bins,rwidth=0.8,histtype='bar',color=['Red','Blue'],label=['RCB','MI'])
plt.title('Franchise Fans')
plt.xlabel('Age Category')
plt.ylabel('No. of Fans')
plt.show()
```

**Pie Chart**

```python
fields=['AI','DS','DA','ML','SE']                                      #Creating field list
Div_a=[12,13,2,4,5]
Div_b=[51,27,37,23,43]
Div_c=[14,10,4,3,2]
```

```python
plt.pie(Div_a,labels=fields,colors=['r','g','b','c','y'])
plt.show()
```

```python
#Advanced Pie Chart                                                                        #Creating Pie Chart Using Styleing
plt.pie(Div_a,labels=fields,explode=[0,0,0,0,0]
        ,colors=['r','g','b','c','y'],autopct='%.1f%%'
        ,shadow=True,startangle=180
        ,textprops={'fontsize':14}
        ,wedgeprops={'linewidth':3,'width':1,'edgecolor':'black'})
plt.title('Pie Chart')
plt.show()
```

```python
#Advanced Pie Chart                                                                        #Creating Pie Chart Using Styleing
plt.pie(Div_a,labels=fields,explode=[0,1,0,0,0]
        ,colors=['r','g','b','c','y'],autopct='%.1f%%'
        ,shadow=True,startangle=180
        ,textprops={'fontsize':14}
        ,wedgeprops={'linewidth':3,'width':1,'edgecolor':'black'})
plt.title('Pie Chart')
plt.show()
```

**Sub Plot,Legend,background Style,Figure Size**

```python
import matplotlib.style as style
style.use('ggplot')
```

```python
plt.figure(figsize=(20,6))
plt.subplot(1,2,1)
plt.plot(pos,run,color='r', marker='o', linestyle='--', markersize=8)
plt.plot(pos,run)
plt.axis([0,11,0,130])
plt.xlabel('Position',fontsize=16)
plt.ylabel('Runs',fontsize=16)
plt.title('Position Wise Scores',fontsize=20)
plt.legend('Scores',loc=1)

plt.subplot(1,2,2)

plt.pie(Div_a,labels=fields,explode=[0,0,0,0,0]
        ,colors=['r','g','b','c','y'],autopct='%.1f%%'
        ,shadow=True,startangle=180
        ,textprops={'fontsize':14}
        ,wedgeprops={'linewidth':3,'width':1,'edgecolor':'black'})
plt.title('Pie Chart')
plt.legend(loc=3)
plt.show()
```
